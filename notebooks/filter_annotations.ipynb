```python
import math
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
ANN_FOLDER = "test_files/IMR90_annotations"
```

```python
signif_thr = -math.log(0.05, 10)
signif_thr
```

```python
for file in os.listdir(ANN_FOLDER):
    print(file)
    df = pd.read_csv(os.path.join(ANN_FOLDER, file), delimiter="\t", header=None)
    df.rename(columns={1: "start", 2: "end", 3: "peak", 4: "score", 8: "qval"}, inplace=True)
    df = df[df["qval"] > signif_thr]
    df = df.nlargest(50_000, "qval")
    df.to_csv(os.path.join(ANN_FOLDER, "top_50000_" + file), sep="\t", header=False, index=False)
    
```
