```python
import pandas as pd
import upsetplot as up
from matplotlib import pyplot
```

```python
pd.set_option('display.max_rows', 500)
```

```python
df = pd.read_csv("permutation_results_0.15_OPT/plots_Rao_2014_IMR90_full_betweenness/pvals_betweenness.csv", index_col=0)
df
```

```python
sig_df = df[(df["a != b"] < 0.05)|(df["a > b"] < 0.05) | (df["a < b"] < 0.05)]
print(f"Number of rows with at least one significant test: {len(sig_df)}")
sig_df
```

```python
sig_df.loc[:, ["a > b"]] = [1 if val  < 0.05 else 0 for val in sig_df["a > b"]]
sig_df.loc[:, ["a < b"]] = [1 if val  < 0.05 else 0 for val in sig_df["a < b"]]
sig_df.loc[:, ["a != b"]] = [1 if val  < 0.05 else 0 for val in sig_df["a != b"]]
sig_df.groupby(["a", "b"]).sum()
```

```python
sig_df[(sig_df["a"] == "1_TssA") & (sig_df["b"] == "Bound") & (sig_df["a < b"] == 1)].sort_values("chr")
```

```python
sig_df[(sig_df["a"] == "7_Enh") & (sig_df["b"] == "Bound") & (sig_df["a < b"] == 1)].sort_values("chr")
```

```python
bigger = sig_df.groupby("chr")["a > b"].count()
smaller = sig_df.groupby("chr")["a < b"].count()
differ = sig_df.groupby("chr")["a != b"].count()

chr_df = pd.DataFrame([bigger, smaller, differ]).T
chr_df
```

```python
sig_df.sort_values("chr")[0:63]
```

```python
upset_df = sig_df.set_index(["a != b", "a < b", "a > b", "a", "b"])
upset_df
```

```python
up.plot(upset_df)  
pyplot.show()  
```
